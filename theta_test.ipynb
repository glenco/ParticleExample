```python
import numpy as np
#from scipy.stats import norm
#from numpy.random import multivariate_normal as multinorm
import matplotlib.pyplot as plt
#from matplotlib import cm
import pandas as pa
from itertools import groupby
from scipy.stats import gamma
import os
import seaborn as sb
```

```python
df = pa.read_csv('theta_test.csv',comment='#')
```

```python
print(len(df['theta']))
```

```python

figure, axis = plt.subplots(2, 2)

for i in range(5):
    
    y=abs( (df['gamma1'+str(i)] - df['gamma1'+str(i)][0] )/df['gamma1'+str(i)][0] )
    axis[0,0].plot(df['theta'],y,label=str(i))
 
#axis[0,0].set_xlabel(r'$\theta_F$')
#axis[0.0].set_ylabel(r'$| [\gamma_1(\theta) - \gamma_1(0) ] / \gamma_1(0) |$')
axis[0,0].set_title(r'$|\Delta\gamma_1(\theta)|$')
axis[0,0].set_ylim([0,0.02])
axis[0,0].set_xlim([0,0.5])

for i in range(5):
    
    y=abs( (df['gamma2'+str(i)] - df['gamma2'+str(i)][0] )/df['gamma2'+str(i)][0] )
    axis[0,1].plot(df['theta'],y,label=str(i))
 
#axis[0,0].set_xlabel(r'$\theta_F$')
#axis[0.0].set_ylabel(r'$| [\gamma_1(\theta) - \gamma_1(0) ] / \gamma_1(0) |$')
axis[0,1].set_title(r'$|\Delta\gamma_2(\theta)|$')
axis[0,1].set_ylim([0,0.02])
axis[0,1].set_xlim([0,0.5])

for i in range(5):
    
    y=abs( (df['alpha_x'+str(i)] - df['alpha_x'+str(i)][0] )/df['alpha_x'+str(i)][0] )
    axis[1,0].plot(df['theta'],y,label=str(i))
 
axis[1,0].set_xlabel(r'$\theta_F$')
#axis[0.0].set_ylabel(r'$| [\gamma_1(\theta) - \gamma_1(0) ] / \gamma_1(0) |$')
axis[1,0].set_title(r'$|\Delta\alpha_1(\theta)|$')
axis[1,0].set_ylim([0,0.02])
axis[1,0].set_xlim([0,0.5])

for i in range(5):
    
    y=abs( (df['alpha_y'+str(i)] - df['alpha_y'+str(i)][0] )/df['alpha_y'+str(i)][0] )
    axis[1,1].plot(df['theta'],y,label=str(i))
 
axis[1,1].set_xlabel(r'$\theta_F$')
#axis[0.0].set_ylabel(r'$| [\gamma_1(\theta) - \gamma_1(0) ] / \gamma_1(0) |$')
axis[1,1].set_title(r'$|\Delta\alpha_2(\theta)|$')
axis[1,1].set_ylim([0,0.02])
axis[1,1].set_xlim([0,0.5])

plt.show()
```
